```python
from PIL import Image, ImageDraw

# Path to the input and output image
input_path = "/Users/qym/Documents/Website/qym7.github.io/images/my_photo.jpg"  # replace with your actual path
output_path = "/Users/qym/Documents/Website/qym7.github.io/images/my_photo.png"


# Open the image and convert to RGBA (to allow transparency)
img = Image.open(input_path).convert("RGBA")

# Get image size
width, height = img.size
crop_size = min(width, height)  # square crop size

# Calculate center crop coordinates
left = (width - crop_size) / 2
top = (height - crop_size) / 2
right = (width + crop_size) / 2
bottom = (height + crop_size) / 2
img_cropped = img.crop((left, top, right, bottom))

# Create a circular mask (L mode = grayscale)
mask = Image.new("L", (crop_size, crop_size), 0)
draw = ImageDraw.Draw(mask)
draw.ellipse((0, 0, crop_size, crop_size), fill=255)

# Apply the mask to create transparency outside the circle
img_circular = Image.new("RGBA", (crop_size, crop_size), (0, 0, 0, 0))
img_circular.paste(img_cropped, (0, 0), mask)

# Save the result as PNG (to keep transparent background)
img_circular.save(output_path)

print(f"Circle crop saved to: {output_path}")
```
